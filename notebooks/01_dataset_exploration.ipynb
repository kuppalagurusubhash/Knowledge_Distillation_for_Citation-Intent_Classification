# 01 - SciCite Dataset Exploration and Preprocessing Analysis
This notebook inspects the SciCite dataset schema, split counts, and class imbalance.


```python
import json
from datasets import load_dataset
import pandas as pd
import matplotlib.pyplot as plt

raw_data = load_dataset('scicite', trust_remote_code=True)
print(raw_data)
```

```python
train_df = raw_data['train'].to_pandas()
print('Label distribution (0=method, 1=background, 2=result):')
print(train_df['label'].value_counts(normalize=True))
```

```python
plt.figure(figsize=(6, 4))
train_df['label'].value_counts().plot(kind='bar', color=['#3B82F6', '#10B981', '#F59E0B'])
plt.title('SciCite Class Distribution (Train)')
plt.xticks([0, 1, 2], ['Background', 'Method', 'Result'], rotation=0)
plt.ylabel('Number of Citations')
plt.show()
```
