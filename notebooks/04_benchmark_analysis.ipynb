Evaluates latency, throughput, model sizes, and robustness.


```python
import json
import pandas as pd
import plotly.express as px

with open('results/benchmarks_summary.json') as f:
    data = json.load(f)

df = pd.DataFrame(data['models'])
display(df[['model_alias', 'method', 'test_macro_f1', 'latency_single_ms', 'throughput_single_eps']])
```

```python
fig = px.scatter(df, x='latency_single_ms', y='test_macro_f1', color='model_alias', title='Latency vs Macro-F1')
fig.show()
```
